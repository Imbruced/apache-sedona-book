```python
from sedona.spark import SedonaContext

config = SedonaContext \
    .builder() \
    .appName('WherobotsExample') \
    .getOrCreate()

sedona = SedonaContext.create(config)
```

```python
sydney_wkt = "POLYGON ((150.5209 -34.1183, 151.3430 -34.1183, 151.3430 -33.5781, 150.5209 -33.5781, 150.5209 -34.1183))"
schema = "overture_maps_foundation"
catalog = "wherobots_open_data"

buildings = sedona.sql(
    f"""
        SELECT *
        FROM {catalog}.{schema}.buildings_building
        WHERE ST_Intersects(geometry, ST_GeomFromText("{sydney_wkt}"))
    """
)
```

```python
complex_buildings = buildings \
    .selectExpr("id", "class", "geometry") \
    .where("ST_NPoints(geometry) > 30")
```

```python
from sedona.spark import SedonaKepler
SedonaKepler.create_map(complex_buildings)
```
