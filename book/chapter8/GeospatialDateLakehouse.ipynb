```python
from sedona.spark import SedonaContext
from sedona.spark.maps.SedonaKepler import SedonaKepler
```

```python
config = SedonaContext.builder()

sedona = SedonaContext.create(config.getOrCreate())
```

# vizualization of area of analysis

```python
nl_shape = sedona.table("sedona_catalog.stg_risk_management.nl_shape").\
    selectExpr("ST_GeomFromEWKB(geometry) AS geometry")

SedonaKepler.create_map(nl_shape, "NL Shape")
```

# vizualization of roads

```python
nl_transportation = sedona.table("sedona_catalog.stg_risk_management.nl_transportation").\
    selectExpr("id", "ST_GeomFromEWKB(geometry) AS geometry")

SedonaKepler.create_map(nl_transportation, "NL roads")
```

# vizualization of NL buildings

```python
nl_buildings = sedona.table("sedona_catalog.stg_risk_management.nl_buildings").\
    selectExpr("id", "ST_GeomFromEWKB(geometry) AS geometry").\
    sample(0.1)

SedonaKepler.create_map(nl_buildings, "Sedona Buildings")
```

# vizualization of nl_transportation_hex_bins

```python
nl_transportation_hex_bins = sedona.table("sedona_catalog.stg_risk_management.nl_transportation_hex_bins").\
    selectExpr( "ST_GeomFromEWKB(geometry) AS geometry")

SedonaKepler.create_map(nl_transportation_hex_bins, "H3 indexes")
```

# vizualization of road risk score index

```python
nl_road_risk_score = sedona.table("sedona_catalog.stg_risk_management.nl_road_risk_score").\
    selectExpr( "ST_GeomFromEWKB(geometry) AS geometry", "RISK_SCORE")

SedonaKepler.create_map(nl_road_risk_score, "Road risk score")
```
